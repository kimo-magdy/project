```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

file_name = "social media and the side effects  (Responses) - Form Responses 1.csv"
df = pd.read_csv(file_name)

df.info()
```

```python
df.head(10)
```

```python
df.columns = df.columns.str.strip()

df.drop(columns=['Email Address'], inplace= True)

df = df.rename(columns={
    'your Current Age': 'Age',
    'Your Gender': 'Gender',
    'Which Social Media platform you frequently use': 'Platforms',
    'what is the usual content that u see': 'Content_Type',
    'your average screen time per day': 'Screen_Time',
    'Did you regret using social media at any point in your day': 'Regret_Usage',
    'Why do you open the social media most of the time': 'Reason_For_Opening',
    'Did the number of likes ever effect your mode': 'Likes_Affect_Mood',
    'From scale 1 - 10 how social media has a bad effect on your focus': 'Bad_Focus_Effect_Scale',
    'From scale 1 - 10 how good your focus without using social media': 'Good_Focus_Without_SM_Scale',
    'Did you ever compare yourself with anyone on social media': 'Compare_With_Others',
    'From scale 1 - 10 how the   comparison often effect your mode in negative way': 'Negative_Comparison_Scale'
})
```

```python
df.head(100)
```

```python
print("--- Descriptive Statistics ---")

numerical_cols = [
    'Bad_Focus_Effect_Scale', 
    'Good_Focus_Without_SM_Scale', 
    'Negative_Comparison_Scale'
]

for col in numerical_cols:
    print(f"\nStatistics for: {col}")
    print(f"Mean:   {df[col].mean():.2f}")
    print(f"Median: {df[col].median()}")
   
    mode_val = df[col].mode()
    if not mode_val.empty:
        print(f"Mode:   {mode_val[0]}")
        
    print("------------------------------")
```

```python
plt.figure(figsize=(14, 5))

plt.subplot(1, 2, 1)
sns.countplot(data=df, x='Age', hue='Gender', palette='pastel')
plt.title('Age and Gender Distribution')
plt.xlabel('Age Group')
plt.ylabel('Count')

plt.subplot(1, 2, 2)

order = ['0 - 3 ( hours )', '3 - 7 ( hours )', '7+ ( hours )']
sns.countplot(data=df, x='Screen_Time', order=order, palette='Set2')
plt.title('Average Daily Screen Time')
plt.xlabel('Screen Time per Day')
plt.ylabel('Count')

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 6))

focus_df = df[['Bad_Focus_Effect_Scale', 'Good_Focus_Without_SM_Scale']].melt(
    var_name='Focus_Metric', value_name='Scale (1-10)'
)
focus_df['Focus_Metric'] = focus_df['Focus_Metric'].replace({
    'Bad_Focus_Effect_Scale': 'Bad Effect on Focus',
    'Good_Focus_Without_SM_Scale': 'Good Focus Without SM'
})

sns.countplot(data=focus_df, x='Scale (1-10)', hue='Focus_Metric', palette='Set2')

plt.title('Impact of Social Media on Focus: Rating Distribution (Scale 1-10)')
plt.ylabel('Number of Respondents')
plt.xlabel('Rating (1 = Lowest, 10 = Highest)')
plt.legend(title='Focus Metric')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(14, 5))

plt.subplot(1, 2, 1)
sns.countplot(data=df, x='Compare_With_Others', palette='coolwarm')
plt.title('Frequency of Social Media Comparisons')
plt.xlabel('Do you compare yourself with others?')
plt.ylabel('Number of Respondents')

plt.subplot(1, 2, 2)
sns.histplot(df['Negative_Comparison_Scale'], bins=10, kde=True, color='red')
plt.title('How Negatively Comparisons Affect Mood (Scale 1-10)')
plt.xlabel('Negative Impact Scale')
plt.ylabel('Frequency')

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

platforms_clean = df['Platforms'].dropna()
platforms_expanded = platforms_clean.str.split(',').explode().str.strip()
platforms_expanded = platforms_expanded.str.lower()
platforms_expanded = platforms_expanded.str.replace('instegram', 'instagram')
platforms_expanded = platforms_expanded.str.title()
platform_counts = platforms_expanded.value_counts()

sns.barplot(x=platform_counts.values, y=platform_counts.index, palette='viridis')
plt.title('Most Frequently Used Social Media Platforms ')
plt.xlabel('Number of Users')
plt.ylabel('Platform')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(9, 5))

sns.countplot(data=df, y='Content_Type', palette='magma', order=df['Content_Type'].value_counts().index)
plt.title('Types of Content Usually Consumed')
plt.xlabel('Number of Users')
plt.ylabel('Content Type')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

order = ['0 - 3 ( hours )', '3 - 7 ( hours )', '7+ ( hours )']

sns.countplot(data=df, x='Screen_Time', hue='Regret_Usage', order=order, palette='Set1')
plt.title('Regret Using Social Media vs. Daily Screen Time')
plt.xlabel('Average Screen Time per Day')
plt.ylabel('Number of Respondents')
plt.legend(title='Do you regret using it?')
plt.tight_layout()
plt.show()
```
